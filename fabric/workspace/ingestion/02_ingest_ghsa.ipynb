````python
import json
import hashlib
from datetime import datetime, timezone

GHSA_URL = "https://api.github.com/advisories"

BRONZE_DIR = "Files/bronze/02_ghsa"
WATERMARK_PATH = f"{BRONZE_DIR}/_watermark.json"

SCHEMA_VERSION = "bronze-v1"
SOURCE = "ghsa"
````

## [2] GHSA 호출 함수 정의

````python
# ======================================
# 공통 Helper 함수
#  - Pagination과 Watermark 범위 생성
# ======================================= 
def to_github_datetime(dt):
    """ 
    GitHub API에 넣을 문자열
    양쪽 경계가 포함될 수 있으므로 경계 중복은 허용하고 누락방지는 Silverr의 ghsa_id + record_hash가 맡는다.
    """
    return dt.strftime("%Y-%m-%dT%H:%M:%SZ")



def _get_next_link(response):
    """
    GitHub Link 헤더에서 rel="next" URL을 추출한다.
    다음 페이지가 없으면 None을 반환한다.
    """
    link_header = response.headers.get("Link")

    if not link_header:
        return None

    for part in link_header.split(","):
        section = part.strip()

        if 'rel="next"' in section:
            return section.split(";")[0].strip().strip("<>")

    return None
````

````python
# ============
# 워터마크 함수
# 운영 실행에서는 Watermark가 없다고 임의로 최근 24시간을 계산하지 않는다.
# 초기 Snapshot은 이미 생성했으므로, Watermark가 없다면 자동 Full Load보다는 실패시켜 상태를 확인하도록 하는 편이 안전하다.
# ===========
def save_ghsa_watermark(watermark_dt):
    """
    Batch 전체 성공 후 새로운 Watermark 저장하는 함수
    """
    watermark_data = {
        "source": "ghsa",
        "last_successful_watermark": to_github_datetime(watermark_dt),
        "updated_at": datetime.now(
            timezone.utc
        ).isoformat().replace("+00:00", "Z")
    }

    watermark_text = json.dumps(
        watermark_data,
        ensure_ascii=False,
        indent=2
    )

    notebookutils.fs.put(
        WATERMARK_PATH,
        watermark_text,
        overwrite=True
    )

    print(
        "watermark saved:",
        watermark_data["last_successful_watermark"]
    )

def load_ghsa_watermark():
    """
    마지막 성공한 워터마크를 읽는 함수
    """
    local_path = (
        f"/lakehouse/default/{WATERMARK_PATH}"
    )

    with open(
        local_path,
        "r",
        encoding="utf-8"
    ) as f:
        watermark_data = json.load(f)

    watermark_str = (
        watermark_data["last_successful_watermark"]
    )

    watermark_dt = datetime.fromisoformat(
        watermark_str.replace("Z", "+00:00")
    )

    return watermark_dt
````

````python
# ========================================
# GHSA 수집 함수
# ========================================
def download_ghsa(modified_range=None):
    """
    GHSA Advisory를 Pagination 끝까지 수집한다.

    modified_range=None
        -> Full Snapshot

    modified_range="start..end"
        -> Incremental
    """

    headers = {
        "Accept": "application/vnd.github+json",
        "Authorization": "Bearer github_pat",
        "X-GitHub-Api-Version": "2022-11-28"
    }

    params = {
        "per_page": 100
    }

    if modified_range:
        params["modified"] = modified_range

    all_advisories = []
    page_count = 0
    
    # next_url을 직접 계산하지 않고 GitHub가 응답한 Link → rel="next"를 그대로 따라가게 한다. 
    next_url = GHSA_URL
    first_request = True

    while next_url:
        if first_request:
            response = request_with_retry(
                "GET",
                next_url,
                headers=headers,
                params=params,
                source="ghsa"
            )
            first_request = False

        else:
            response = request_with_retry(
                "GET",
                next_url,
                headers=headers,
                source="ghsa"
            )

        page_data = response.json()

        if not isinstance(page_data, list):
            raise ValueError(
                f"GHSA response is not a list: "
                f"type={type(page_data)}"
            )

        page_count += 1
        all_advisories.extend(page_data)

        print(
            f"page={page_count}, "
            f"page_records={len(page_data)}, "
            f"total_records={len(all_advisories)}"
        )

        next_url = _get_next_link(response)

    return all_advisories, page_count
````

````python
# =====================================
# 수집결과 최종 검증 함수
# ====================================
def validate_ghsa(records, allow_empty=False):
    """GHSA ID 기준 중복·필수값·응답 타입 검증 함수"""

    if not isinstance(records, list):
        raise TypeError(
            f"GHSA records must be list, got {type(records).__name__}"
        )

    # 증분 수집에서는 0건을 정상 no-op으로 허용: 0건이어도 해당 watermark_start → watermark_end 구간을 정상 확인했다면 watermark는 앞으로 이동 가능
    if len(records) == 0:
        if allow_empty:
            result = {
                "record_count": 0,
                "invalid_record_type": 0,
                "missing_ghsa_id": 0,
                "invalid_ghsa_id_type": 0,
                "duplicate_ghsa_id": 0,
                "status": "no_op",
            }

            print(result)
            return result

        raise ValueError("GHSA records is empty")

    invalid_record_type = 0
    invalid_ghsa_id_type = 0
    missing_ghsa_id = 0

    ghsa_ids = []

    for record in records:
        if not isinstance(record, dict):
            invalid_record_type += 1
            continue

        ghsa_id = record.get("ghsa_id")

        if ghsa_id is None or ghsa_id == "":
            missing_ghsa_id += 1
            continue

        if not isinstance(ghsa_id, str):
            invalid_ghsa_id_type += 1
            continue

        ghsa_ids.append(ghsa_id)

    duplicate_count = len(ghsa_ids) - len(set(ghsa_ids))

    result = {
        "record_count": len(records),
        "invalid_record_type": invalid_record_type,
        "missing_ghsa_id": missing_ghsa_id,
        "invalid_ghsa_id_type": invalid_ghsa_id_type,
        "duplicate_ghsa_id": duplicate_count,
        "status": "success",
    }

    print(result)

    if (
        invalid_record_type > 0
        or missing_ghsa_id > 0
        or invalid_ghsa_id_type > 0
        or duplicate_count > 0
    ):
        raise ValueError(f"GHSA validation failed: {result}")

    return result
````

````python
# ===============
# JSONL + Metadata 생성 함수
# =================

def build_ghsa_jsonl(records):
    """
    GHSA Advisory records를 JSONL 문자열로 변환하고
    SHA-256 checksum을 생성한다.
    """

    if not isinstance(records, list):
        raise TypeError("records must be a list")

    jsonl_text = "\n".join(
        json.dumps(
            record,
            ensure_ascii=False,
            separators=(",", ":")
        )
        for record in records
    ) + "\n"

    jsonl_bytes = jsonl_text.encode("utf-8")

    checksum = hashlib.sha256(
        jsonl_bytes
    ).hexdigest()

    return jsonl_text, jsonl_bytes, checksum

def build_ghsa_metadata(
    records,
    checksum,
    page_count,
    previous_watermark,
    batch_window_end
):
    """
    GHSA Incremental Batch의 파일 경로와
    Bronze metadata를 생성한다.
    """

    batch_ts = batch_window_end.strftime("%Y%m%dT%H%M%SZ")
    batch_id = f"ghsa_{batch_ts}"

    raw_file_name = f"{batch_id}.jsonl"
    meta_file_name = f"{batch_id}_meta.json"

    raw_path = f"{BRONZE_DIR}/{raw_file_name}"
    meta_path = f"{BRONZE_DIR}/{meta_file_name}"

    fetched_at = datetime.now(
        timezone.utc
    ).isoformat().replace("+00:00", "Z")

    metadata = {
        "source": SOURCE,
        "batch_id": batch_id,
        "fetched_at": fetched_at,
        "file_name": raw_file_name,
        "format": "jsonl",
        "record_count": len(records),
        "checksum": f"sha256:{checksum}",
        "schema_version": SCHEMA_VERSION,

        "source_url": GHSA_URL,
        "page_count": page_count,
        
        # 아래 세 가지는 공통 필수 계약은 아니지만 gsha 증분 수집의 리니지와 재실행 범위를 설명하기 위해 필요한 메타데이터
        "collection_mode": "incremental",
        "watermark_start": to_github_datetime(
            previous_watermark
        ),
        "watermark_end": to_github_datetime(
            batch_window_end
        )
    }

    return metadata, raw_path, meta_path
````

````python
# =======================
# bronze에 저장
# =========================
def save_ghsa_bronze(
    jsonl_text,
    metadata,
    raw_path,
    meta_path
):
    """
    GHSA Raw JSONL과 대응하는 metadata를
    Bronze에 한 쌍으로 저장한다.

    기존 Bronze 파일은 덮어쓰지 않는다.
    """

    meta_text = json.dumps(
        metadata,
        ensure_ascii=False,
        indent=2
    )

    notebookutils.fs.put(
        raw_path,
        jsonl_text,
        overwrite=False
    )

    notebookutils.fs.put(
        meta_path,
        meta_text,
        overwrite=False
    )

    print("GHSA Bronze 저장 완료")
    print("raw :", raw_path)
    print("meta:", meta_path)

    return raw_path, meta_path
````

````python
# ==========================
# 해당 함수가 성공하면 watermark commit
# =============================
def verify_ghsa_bronze(
    expected_record_count,
    expected_checksum,
    raw_path,
    meta_path
):
    """
    Bronze 저장 결과를 검증한다.

    검증 항목
    1. Source record count == Bronze JSONL record count
    2. 저장 전 checksum == 저장 후 checksum
    3. metadata record_count == 실제 Bronze record count
    4. metadata checksum == 실제 Bronze checksum

    하나라도 실패하면 예외를 발생시킨다.
    """

    local_raw_path = f"/lakehouse/default/{raw_path}"
    local_meta_path = f"/lakehouse/default/{meta_path}"

    # -------------------------
    # 1. 저장된 Raw checksum
    # -------------------------
    saved_hash = hashlib.sha256()

    with open(local_raw_path, "rb") as f:
        for chunk in iter(
            lambda: f.read(1024 * 1024),
            b""
        ):
            saved_hash.update(chunk)

    saved_checksum = saved_hash.hexdigest()

    # -------------------------
    # 2. 저장된 Raw record count
    # -------------------------
    with open(
        local_raw_path,
        "r",
        encoding="utf-8"
    ) as f:
        bronze_record_count = sum(
            1
            for line in f
            if line.strip()
        )

    # -------------------------
    # 3. 저장된 metadata
    # -------------------------
    with open(
        local_meta_path,
        "r",
        encoding="utf-8"
    ) as f:
        saved_meta = json.load(f)

    # -------------------------
    # 4. 비교
    # -------------------------
    record_count_match = (
        expected_record_count
        == bronze_record_count
    )

    checksum_match = (
        expected_checksum
        == saved_checksum
    )

    meta_record_count_match = (
        saved_meta["record_count"]
        == bronze_record_count
    )

    meta_checksum_match = (
        saved_meta["checksum"]
        == f"sha256:{saved_checksum}"
    )

    validation_result = {
        "expected_record_count": expected_record_count,
        "bronze_record_count": bronze_record_count,
        "record_count_match": record_count_match,
        "checksum_match": checksum_match,
        "meta_record_count_match": meta_record_count_match,
        "meta_checksum_match": meta_checksum_match
    }

    # 하나라도 실패하면 Batch 실패
    if not all([
        record_count_match,
        checksum_match,
        meta_record_count_match,
        meta_checksum_match
    ]):
        raise ValueError(
            f"GHSA Bronze validation failed: "
            f"{validation_result}"
        )

    print("GHSA Bronze 검증 성공")
    print(
        json.dumps(
            validation_result,
            ensure_ascii=False,
            indent=2
        )
    )

    return validation_result
````

## [3] 실제 일일 실행 순서 

````python
# ============================================================
# GHSA Daily Incremental Ingestion - Production Run
# ============================================================

run_started_at = datetime.now(timezone.utc)

print("========================================")
print("GHSA Incremental Ingestion START")
print("run_started_at:", to_github_datetime(run_started_at))
print("========================================")


# ------------------------------------------------------------
# 1. 마지막 성공 Watermark 읽기
# ------------------------------------------------------------
previous_watermark = load_ghsa_watermark()

# 이번 Batch가 책임질 종료 시각을 수집 시작 전에 고정
batch_window_end = datetime.now(timezone.utc)

modified_range = (
    f"{to_github_datetime(previous_watermark)}"
    f".."
    f"{to_github_datetime(batch_window_end)}"
)

print("watermark_start:", to_github_datetime(previous_watermark))
print("watermark_end  :", to_github_datetime(batch_window_end))
print("modified_range :", modified_range)


# ------------------------------------------------------------
# 2. GHSA 증분 수집
# ------------------------------------------------------------
ghsa_records, ghsa_page_count = download_ghsa(
    modified_range=modified_range
)

print("pagination complete")
print("pages          :", ghsa_page_count)
print("source records :", len(ghsa_records))


# ------------------------------------------------------------
# 3. 수집 데이터 기본 검증
# ------------------------------------------------------------
validation_result = validate_ghsa(
    ghsa_records,
    allow_empty=True
)


# ------------------------------------------------------------
# 4. No-op / 정상 적재 분기
# ------------------------------------------------------------
if len(ghsa_records) == 0:

    # --------------------------------------------------------
    # 4-1. 변경 없음(No-op)
    # --------------------------------------------------------
    print("========================================")
    print("GHSA Incremental Ingestion NO-OP")
    print("변경된 GHSA Advisory가 없습니다.")
    print("Bronze Raw 저장을 생략합니다.")

    # 해당 구간을 정상적으로 확인했으므로 Watermark Commit
    save_ghsa_watermark(
        batch_window_end
    )

    run_finished_at = datetime.now(timezone.utc)

    print("watermark_start :", to_github_datetime(previous_watermark))
    print("watermark_end   :", to_github_datetime(batch_window_end))
    print("page_count      :", ghsa_page_count)
    print("record_count    :", 0)
    print("status          : no_op")
    print("run_finished_at :", to_github_datetime(run_finished_at))
    print("========================================")


else:

    # --------------------------------------------------------
    # 4-2. Bronze JSONL + checksum 생성
    # --------------------------------------------------------
    jsonl_text, jsonl_bytes, checksum = build_ghsa_jsonl(
        ghsa_records
    )


    # --------------------------------------------------------
    # 5. Metadata + 저장 경로 생성
    # --------------------------------------------------------
    metadata, raw_path, meta_path = build_ghsa_metadata(
        records=ghsa_records,
        checksum=checksum,
        page_count=ghsa_page_count,
        previous_watermark=previous_watermark,
        batch_window_end=batch_window_end
    )


    # --------------------------------------------------------
    # 6. Bronze Raw + Metadata 저장
    # --------------------------------------------------------
    save_ghsa_bronze(
        jsonl_text=jsonl_text,
        metadata=metadata,
        raw_path=raw_path,
        meta_path=meta_path
    )


    # --------------------------------------------------------
    # 7. Source ↔ Bronze Reconciliation
    # --------------------------------------------------------
    bronze_validation = verify_ghsa_bronze(
        expected_record_count=len(ghsa_records),
        expected_checksum=checksum,
        raw_path=raw_path,
        meta_path=meta_path
    )


    # --------------------------------------------------------
    # 8. 모든 단계 성공 후에만 Watermark Commit
    # --------------------------------------------------------
    save_ghsa_watermark(
        batch_window_end
    )


    # --------------------------------------------------------
    # 9. 최종 실행 결과
    # --------------------------------------------------------
    run_finished_at = datetime.now(timezone.utc)

    print("========================================")
    print("GHSA Incremental Ingestion SUCCESS")
    print("batch_id           :", metadata["batch_id"])
    print("watermark_start    :", metadata["watermark_start"])
    print("watermark_end      :", metadata["watermark_end"])
    print("page_count         :", ghsa_page_count)
    print("source_record_count:", len(ghsa_records))
    print(
        "bronze_record_count:",
        bronze_validation["bronze_record_count"]
    )
    print(
        "checksum_match     :",
        bronze_validation["checksum_match"]
    )
    print("raw_path           :", raw_path)
    print("meta_path          :", meta_path)
    print("status             : success")
    print("run_finished_at    :", to_github_datetime(run_finished_at))
    print("========================================")
````
